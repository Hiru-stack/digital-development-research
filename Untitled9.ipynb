```python
import pandas as pd
import numpy as np

df = pd.read_csv("itu_extract_1790938717412.csv")

print(df.shape)
df.head()
```

```python
countries = [
    "Sri Lanka",
    "India",
    "Bangladesh",
    "Nepal",
    "Indonesia",
    "Malaysia"
]

asia_df = df[df["AreaName"].isin(countries)].copy()

print(asia_df.shape)
print(asia_df["AreaName"].unique())
```

```python
print(sorted(asia_df["Year"].unique()))
```

```python
for indicator in sorted(asia_df["Series"].dropna().unique()):
    print(indicator)
```

```python
internet = asia_df[
    asia_df["Series"] == "Internet users (%)"
].copy()

internet = internet[
    ["AreaName", "Year", "Value"]
]

internet = internet.rename(columns={
    "AreaName": "Country",
    "Value": "Internet_Users_Percent"
})

internet.head(20)
```

```python
internet.info()
```

```python
internet["Internet_Users_Percent"] = pd.to_numeric(
    internet["Internet_Users_Percent"],
    errors="coerce"
)
```

```python
internet.isnull().sum()
```

```python
internet.groupby("Country")["Internet_Users_Percent"].count()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,7))

for country in countries:
    temp = internet[internet["Country"] == country]

    plt.plot(
        temp["Year"],
        temp["Internet_Users_Percent"],
        marker="o",
        label=country
    )

plt.xlabel("Year")
plt.ylabel("Individuals Using the Internet (%)")
plt.title("Internet Use in Selected Asian Countries, 2014–2025")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
mobile_bb = asia_df[
    asia_df["Series"] ==
    "Active mobile-broadband subscriptions per 100 inhabitants"
].copy()

mobile_bb = mobile_bb[
    ["AreaName", "Year", "Value"]
]

mobile_bb.columns = [
    "Country",
    "Year",
    "Mobile_Broadband_Per_100"
]

mobile_bb["Mobile_Broadband_Per_100"] = pd.to_numeric(
    mobile_bb["Mobile_Broadband_Per_100"],
    errors="coerce"
)

mobile_bb.head()
```

```python
mobile_bb.groupby("Country")[
    "Mobile_Broadband_Per_100"
].count()
```

```python
plt.figure(figsize=(12,7))

for country in countries:
    temp = mobile_bb[mobile_bb["Country"] == country]

    plt.plot(
        temp["Year"],
        temp["Mobile_Broadband_Per_100"],
        marker="o",
        label=country
    )

plt.xlabel("Year")
plt.ylabel("Subscriptions per 100 inhabitants")
plt.title("Active Mobile Broadband in Selected Asian Countries")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
fixed_bb = asia_df[
    asia_df["Series"] ==
    "Fixed broadband subscriptions per 100 inhabitants"
].copy()

fixed_bb = fixed_bb[
    ["AreaName", "Year", "Value"]
]

fixed_bb.columns = [
    "Country",
    "Year",
    "Fixed_Broadband_Per_100"
]

fixed_bb["Fixed_Broadband_Per_100"] = pd.to_numeric(
    fixed_bb["Fixed_Broadband_Per_100"],
    errors="coerce"
)

fixed_bb.head()
```

```python
combined = internet.merge(
    mobile_bb,
    on=["Country", "Year"],
    how="outer"
)

combined = combined.merge(
    fixed_bb,
    on=["Country", "Year"],
    how="outer"
)

combined = combined.sort_values(
    ["Country", "Year"]
)

combined.head(20)
```

```python
combined.isnull().sum()
```

```python
combined.groupby("Country").count()
```

```python
combined["Internet_Growth"] = (
    combined.groupby("Country")["Internet_Users_Percent"]
    .pct_change() * 100
)
```

```python
combined["Internet_Annual_Change"] = (
    combined.groupby("Country")["Internet_Users_Percent"]
    .diff()
)
```

```python
latest = (
    combined
    .dropna(subset=["Internet_Users_Percent"])
    .sort_values("Year")
    .groupby("Country")
    .tail(1)
)

latest[
    ["Country", "Year", "Internet_Users_Percent"]
].sort_values(
    "Internet_Users_Percent",
    ascending=False
)
```

```python
# 1. Internet users
internet = asia_df[
    asia_df["Series"] == "Internet users (%)"
][["AreaName", "Year", "Value"]].copy()

internet.columns = ["Country", "Year", "Internet_Users_Percent"]

internet["Internet_Users_Percent"] = pd.to_numeric(
    internet["Internet_Users_Percent"],
    errors="coerce"
)
```

```python
# 2. Mobile broadband
mobile_bb = asia_df[
    asia_df["Series"] ==
    "Active mobile-broadband subscriptions per 100 inhabitants"
][["AreaName", "Year", "Value"]].copy()

mobile_bb.columns = [
    "Country",
    "Year",
    "Mobile_Broadband_Per_100"
]

mobile_bb["Mobile_Broadband_Per_100"] = pd.to_numeric(
    mobile_bb["Mobile_Broadband_Per_100"],
    errors="coerce"
)
```

```python
# 3. Fixed broadband
fixed_bb = asia_df[
    asia_df["Series"] ==
    "Fixed broadband subscriptions per 100 inhabitants"
][["AreaName", "Year", "Value"]].copy()

fixed_bb.columns = [
    "Country",
    "Year",
    "Fixed_Broadband_Per_100"
]

fixed_bb["Fixed_Broadband_Per_100"] = pd.to_numeric(
    fixed_bb["Fixed_Broadband_Per_100"],
    errors="coerce"
)
```

```python
itu_clean = itu_compare[
    itu_compare["Year"] <= 2024
].copy()

itu_clean.isnull().sum()
```

```python
itu_clean.groupby("Country").count()
```

```python
internet_change = (
    itu_clean.groupby("Country")
    .agg(
        Start_Internet=("Internet_Users_Percent", "first"),
        End_Internet=("Internet_Users_Percent", "last")
    )
)

internet_change["Increase_pp"] = (
    internet_change["End_Internet"]
    - internet_change["Start_Internet"]
)

internet_change.sort_values(
    "Increase_pp",
    ascending=False
)
```

```python
internet_wide = itu_clean.pivot(
    index="Year",
    columns="Country",
    values="Internet_Users_Percent"
)

for country in ["India", "Bangladesh", "Indonesia", "Malaysia"]:
    internet_wide[f"SL_gap_vs_{country}"] = (
        internet_wide["Sri Lanka"]
        - internet_wide[country]
    )

internet_wide
```

```python
plt.figure(figsize=(11,6))

for country in ["India", "Bangladesh", "Indonesia", "Malaysia"]:
    plt.plot(
        internet_wide.index,
        internet_wide[f"SL_gap_vs_{country}"],
        marker="o",
        label=f"Sri Lanka vs {country}"
    )

plt.axhline(0, linestyle="--")

plt.xlabel("Year")
plt.ylabel("Difference in Internet Use (Percentage Points)")
plt.title("Sri Lanka's Internet-Use Gap with Selected Asian Countries")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
